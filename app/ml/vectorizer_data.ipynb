```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(
    vectorized_df['preprocessed_absract']
)
```

```python
print(tfidf_matrix.shape)
print(tfidf_vectorizer.get_feature_names_out()[:20])
```

```python
import numpy as np
import pandas as pd

feature_names = tfidf_vectorizer.get_feature_names_out()


tfidf_global = tfidf_matrix.sum(axis=0)  # type: ignore # shape: (1, n_features)

tfidf_values = np.asarray(tfidf_global).flatten()

tfidf_df = pd.DataFrame({
    "term": feature_names,
    "tfidf": tfidf_values
}).sort_values(by="tfidf", ascending=False)

tfidf_df.head(1283)
```

```python
from sklearn.metrics.pairwise import cosine_similarity


cosine_sim_matrix = cosine_similarity(tfidf_matrix, tfidf_matrix)



cosine_sim_df = pd.DataFrame(
    cosine_sim_matrix, 
    index=[f"Doc {i}" for i in range(tfidf_matrix.shape[0])],
    columns=[f"Doc {i}" for i in range(tfidf_matrix.shape[0])]
)


print("Matriks Cosine Similarity (Sampel):")
print(cosine_sim_df.iloc[:20, :20])
```

```python
for i in range(10): 
    sims = cosine_sim_matrix[i]
    top_indices = sims.argsort()[::-1][1:6]
    
    print(f"\nDoc {i} paling mirip dengan:")
    for idx in top_indices:
        print(f"  -> Doc {idx} (score: {sims[idx]:.3f})")
```

```python
from cbf import ContentBasedFiltering


cbf = ContentBasedFiltering(datas, tfidf_vectorizer, tfidf_matrix)

input_abstract = datas['abstract'][10]
recommendations = cbf.get_recommendations_by_abstract(
    input_abstract,
    top_n=100
)
```
